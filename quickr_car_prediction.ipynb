```python
X = car_ds.drop(columns=['Price'])
Y = car_ds['Price']
```

```python
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2)
```

```python
ohe = OneHotEncoder()
ohe.fit(X[['name','company','fuel_type']])
```

```python

col_trans = make_column_transformer((OneHotEncoder(categories=ohe.categories_),['name','company','fuel_type']),remainder='passthrough')
```

LASSO REGRESSION

```python
l = Lasso()
```

```python
pl = make_pipeline(col_trans,l)
```

```python
pl.fit(X_train,Y_train)
```

```python
l_y_pred = pl.predict(X_test)
```

```python
(l_y_pred<0).sum()
```

```python
pl.predict(pd.DataFrame([['Maruti Suzuki Swift','Maruti','2026','500000','Petrol']],columns=['name','company','year','kms_driven','fuel_type']))
```

```python
metrics.r2_score(Y_test,l_y_pred)
```

LINEAR REGRESSION



```python
lr = LinearRegression()
```

```python
pipe = make_pipeline(col_trans,lr)
```

```python
pipe.fit(X_train,Y_train)
```

```python
y_pred = pipe.predict(X_test)
```

```python
(Y_test < 0).value_counts()
```

```python
(y_pred <0).sum()
```

```python
metrics.r2_score(Y_test,y_pred)
```
